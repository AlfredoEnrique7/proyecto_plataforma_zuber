En esta sección se automatiza, mediante el codigo que se muestra a continuación, la recolección de las variables externas de la investigación (el historial meteorológico de Chicago de noviembre de 2017) directamente desde el sitio web del proyecto. Esta información servirá como base para evaluar el impacto del clima en la frecuencia y duración de los viajes compartidos.

````python
# ===============================================================
# Se cargan las librerías necesarias para desarrollar el proyecto
# ===============================================================
import requests
from bs4 import BeautifulSoup
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
import seaborn as sns
````

````python
##########################################################
# Descarga y análisis del árbol documental (HTML parsing)
##########################################################
URL='https://practicum-content.s3.us-west-1.amazonaws.com/data-analyst-eng/moved_chicago_weather_2017.html'
req = requests.get(URL)
soup = BeautifulSoup(req.text, 'lxml')
table = soup.find('table', attrs={"id": "weather_records"})

##########################################################
# Extracción estructurada de los componentes
##########################################################
heading_table=[]
for row in table.find_all('th'):
    heading_table.append(row.text)
content=[]
for row in table.find_all('tr'):
    if not row.find_all('th'):
        content.append([element.text for element in row.find_all('td')])

##########################################################
# Consolidación matricial de los datos
##########################################################
weather_records = pd.DataFrame(content, columns = heading_table)
print(weather_records)
````

### 2.1 Cantidad de viajes por compañía de taxis (15 y 16 de Noviembre de 2017)

Esta consulta consolida y clasifica el volumen absoluto de operaciones ejecutadas por cada empresa de taxis en Chicago durante un periodo crítico de dos días a mediados de noviembre de 2017.

Para resolver la primera parte del análisis exploratorio en SQL (Paso 2.1), se debe cruzar la tabla de viajes (`trips`) con la tabla de taxis (`cabs`) a través del campo común cab_id.
Dado que se necesita cuantificar la actividad operativa de cada empresa exclusivamente para los días 15 y 16 de noviembre de 2017, se aplica una función de agregación `COUNT(*)` agrupando por el nombre de la compañía. Utilizaremos el casting de fecha `CAST(start_ts AS date)` para asegurarnos de capturar con exactitud todos los registros de ambas jornadas completas.

```sql
SELECT
    cabs.company_name AS company_name,
    COUNT(trips.trip_id) AS trips_amount
FROM
    trips
    INNER JOIN cabs ON trips.cab_id = cabs.cab_id
WHERE
    CAST(trips.start_ts AS date) BETWEEN '2017-11-15' AND '2017-11-16'
GROUP BY
    cabs.company_name
ORDER BY
    trips_amount DESC;
```

#### Explicación estructurada del código:

1. **Selección y Métricas de Rendimiento (`SELECT`)**:
   * **`cabs.company_name AS company_name`**: Extrae el nombre de la empresa propietaria del vehículo desde la tabla maestra `cabs`.
   * **`COUNT(trips.trip_id) AS trips_amount`**: Cuenta de forma agregada el identificador único de cada viaje para calcular la frecuencia total de viajes por grupo, asignándole el alias obligatorio `trips_amount`.

2. **Cruces Relacionales (`FROM` e `INNER JOIN`)**:
   * Utiliza la tabla operativa `trips` como pivote y realiza un acoplamiento interno (`INNER JOIN`) con la tabla de vehículos `cabs` mediante la clave común `cab_id`, asegurando que cada viaje contabilizado esté perfectamente asociado a una marca corporativa real.

3. **Recorte Temporal de Datos (`WHERE`)**:
   * **`CAST(trips.start_ts AS date) BETWEEN ...`**: Convierte la marca de tiempo precisa (`TIMESTAMP`) a formato de fecha corta para omitir horas intermitentes. El operador `BETWEEN` restringe el procesamiento para incluir únicamente las transacciones transcurridas entre las 00:00:00 del 15 de noviembre y las 23:59:59 del 16 de noviembre de 2017.

4. **Clasificación y Ordenamiento (`GROUP BY` y `ORDER BY`)**:
   * **`GROUP BY cabs.company_name`**: Concentra las filas individuales en contenedores basados en el nombre de la empresa de taxis para calcular de manera independiente su rendimiento operativo.
   * **`ORDER BY trips_amount DESC`**: Ordena los grupos de forma descendente (de la compañía con mayor cantidad de viajes a la menor), permitiendo identificar de forma inmediata a los líderes dominantes del mercado de transporte en Chicago.

### 2.2 Viajes de las compañías "Yellow" o "Blue" (1 al 7 de Noviembre de 2017)

Esta consulta filtra y cuantifica de forma unívoca el desempeño operativo de las flotas de taxis cuyos nombres corporativos incluyen los distintivos colores "Yellow" o "Blue" durante la primera semana de noviembre de 2017.

Para resolver esta consulta, se utiliza el operador `LIKE` para buscar patrones de texto parciales en el nombre de la empresa, combinándolo con operadores lógicos para cumplir las pautas. Restringiremos los datos del **1 al 7 de noviembre de 2017** utilizando BETWEEN en la fecha truncada mediante `CAST`.

```sql
SELECT
    cabs.company_name AS company_name,
    COUNT(trips.trip_id) AS trips_amount
FROM
    trips
    INNER JOIN cabs ON trips.cab_id = cabs.cab_id
WHERE
    (cabs.company_name LIKE '%Yellow%' OR cabs.company_name LIKE '%Blue%')
    AND CAST(trips.start_ts AS date) BETWEEN '2017-11-01' AND '2017-11-07'
GROUP BY
    cabs.company_name;
```
#### Explicación estructurada del código:

1. **Selección y variable agregada (`SELECT`)**:
   * **`cabs.company_name`**: Muestra el nombre comercial de las empresas seleccionadas.
   * **`COUNT(trips.trip_id) AS trips_amount`**: Cuenta la cantidad absoluta de viajes por grupo corporativo, guardando obligatoriamente el resultado bajo la variable `trips_amount`.

2. **Filtro avanzado por palabras clave y fecha (`WHERE`)**:
   * **`LIKE '%Yellow%' OR LIKE '%Blue%'`**: El uso de los comodines `%` a los lados permite buscar cualquier empresa cuyo nombre *contenga* de forma parcial estas palabras tradicionales de taxis de Chicago (ej. *Yellow Cab*, *Blue Ribbon Taxi*, etc.). Los paréntesis son estrictamente necesarios para que el motor SQL evalúe primero el color y luego la fecha.
   * **`CAST(trips.start_ts AS date) BETWEEN ...`**: Homologa las marcas de tiempo a formato de fecha corta y segmenta de forma exclusiva las transacciones ejecutadas del **1 al 7 de noviembre de 2017**.

3. **Agrupamiento requerido (`GROUP BY`)**:

### 2.3 Segmentación de viajes: Líderes del mercado frente a competidores ("Other")

Esta consulta consolida la cuota de mercado en Chicago durante la primera semana de noviembre de 2017, aislando de forma individual el desempeño de las dos empresas más populares (`Flash Cab` y `Taxi Affiliation Services`) y agrupando jerárquicamente a todos los demás competidores en una sola categoría.

Para resolver este ejercicio de forma óptima, utilizaremos la estructura condicional **`CASE WHEN`** dentro de la agregación. Esta función nos permite evaluar el nombre de la empresa para mantener a los líderes `"Flash Cab"` y `"Taxi Affiliation Services"` con su marca original y clasificar de forma masiva a todas las demás empresas competidoras bajo la categoría general `"Other".

```sql
SELECT
    CASE 
        WHEN cabs.company_name = 'Flash Cab' THEN 'Flash Cab'
        WHEN cabs.company_name = 'Taxi Affiliation Services' THEN 'Taxi Affiliation Services'
        ELSE 'Other'
    END AS company,
    COUNT(trips.trip_id) AS trips_amount
FROM
    trips
    INNER JOIN cabs ON trips.cab_id = cabs.cab_id
WHERE
    CAST(trips.start_ts AS date) BETWEEN '2017-11-01' AND '2017-11-07'
GROUP BY
    company
ORDER BY
    trips_amount DESC;
```

#### Explicación estructurada del código:

1. **Lógica condicional de clasificación (`CASE WHEN`)**:
   * Evalúa de forma secuencial la columna `company_name`. Si el nombre coincide exactamente con `'Flash Cab'` o `'Taxi Affiliation Services'`, retiene dicha cadena de texto. Para cualquier otra razón social del catálogo maestro, la instrucción `ELSE` la reclasifica de forma automática como `'Other'`, almacenando el resultado bajo la variable solicitada `company`.

2. **Métricas y relaciones entre tablas**:
   * **`COUNT(trips.trip_id) AS trips_amount`**: Cuantifica de forma agregada el número absoluto de viajes correspondientes a cada nuevo grupo de mercado definido.
   * Cruza la tabla operativa de registros de viajes (`trips`) con la tabla de taxis (`cabs`) mediante el identificador común `cab_id`.

3. **Restricción temporal y agrupamiento compuesto**:
   * **`CAST(trips.start_ts AS date) BETWEEN ...`**: Homologa las marcas de tiempo y segmenta las operaciones transcurridas estrictamente entre el **1 y el 7 de noviembre de 2017**.
   * **`GROUP BY company`**: Agrupa y compacta las filas utilizando el alias de la expresión condicional. Esto permite que la función de conteo sume verticalmente todos los viajes de las compañías más pequeñas dentro del contenedor único `'Other'`.
   * **`ORDER BY trips_amount DESC`**: Ordena la matriz resultante de mayor a menor frecuencia operativa, permitiendo constatar visualmente la dominancia de los gigantes del transporte en Chicago.

### 3.2. Clasificación de condiciones meteorológicas horarias (Good vs. Bad)

Esta consulta procesa el historial por hora de la tabla `weather_records` y categoriza de manera binaria el estado del clima en "Good" (Bueno) o "Bad" (Malo), sentando las bases lógicas para evaluar posteriormente el impacto ambiental en la duración de los viajes de Zuber.

Para resolver este ejercicio, se utilizará el operador condicional `CASE WHEN` junto con el operador de coincidencia parcial `LIKE`. Esto permitirá escanear el campo de descripción climática en busca de los patrones `'rain'` (lluvia) o `'storm'` (tormenta) y clasificar cada hora de forma automatizada.

```sql
SELECT
    ts,
    CASE
        WHEN description LIKE '%rain%' OR description LIKE '%storm%' THEN 'Bad'
        ELSE 'Good'
    END AS weather_conditions
FROM
    weather_records;
```

#### Explicación estructurada del código:

1. **Selección Cronológica Directa**:
   * **`ts`**: Recupera el campo original de fecha y hora del registro meteorológico (redondeado a la hora exacta), el cual servirá en la siguiente etapa como puente de conexión temporal con los viajes.

2. **Lógica de Clasificación Binaria (`CASE WHEN ... LIKE`)**:
   * El operador **`CASE`** evalúa secuencialmente las cadenas de texto del campo `description`.
   * **`LIKE '%rain%' OR LIKE '%storm%'`**: Los comodines porcentuales (`%`) a los extremos permiten identificar cualquier variante descriptiva que contenga de forma parcial las palabras clave de mal clima (ej. *'light rain'*, *'heavy rain'*, *'thunderstorm'*).
   * Si la condición se cumple, la hora recibe la etiqueta **`'Bad'`**; para cualquier otro escenario atmosférico despejado o nublado, la instrucción `ELSE` asigna de forma predeterminada la etiqueta **`'Good'`**. Todo el bloque se guarda bajo el alias requerido **`weather_conditions`**.

3. **Origen de Datos**:
   * **`FROM weather_records`**: Extrae la información directamente de la tabla meteorológica histórica de Chicago.

### 3.3. Dataset consolidado para la prueba de hipótesis (Zuber)

Esta consulta genera el conjunto de datos definitivo uniendo los registros operativos de los viajes del centro al aeropuerto los sábados con las condiciones meteorológicas horarias correspondientes, filtrando los valores ausentes.

Para resolver este ejercicio (el dataset final para la prueba de hipótesis), se une la tabla `trips` con la subconsulta de condiciones climáticas desarrollada en el paso anterior.

Dado que se pide filtrar solo los viajes que comenzaron en **sábado**, se utiliza la función **`EXTRACT(ISODOW FROM start_ts) = 6`** o **`EXTRACT(DOW FROM start_ts) = 6`** (dependiendo del motor de la base de datos de tu plataforma) o la abreviación de texto equivalente. El uso de un **`INNER JOIN`** descartará automáticamente los viajes que no tengan datos climáticos asociados, cumpliendo con la directiva de ignorar registros vacíos.

```sql
SELECT
    trips.start_ts AS start_ts,
    CASE
        WHEN weather_records.description LIKE '%rain%' OR weather_records.description LIKE '%storm%' THEN 'Bad'
        ELSE 'Good'
    END AS weather_conditions,
    trips.duration_seconds AS duration_seconds
FROM
    trips
INNER JOIN 
    weather_records ON trips.start_ts = weather_records.ts
WHERE
    trips.pickup_location_id = 50
    AND trips.dropoff_location_id = 63
    AND EXTRACT(DOW FROM trips.start_ts) = 6
ORDER BY
    trips.trip_id;
```

#### Explicación estructurada del código:

1. **Orden de Columnas Requerido (`SELECT`)**:
   * **`trips.start_ts AS start_ts`**: Registra el momento exacto en el que inició la ruta.
   * **`CASE WHEN ... END AS weather_conditions`**: Integra la segmentación climática binaria de la tarea anterior directamente en el flujo principal del reporte.
   * **`trips.duration_seconds AS duration_seconds`**: Proyecta la variable dependiente de la investigación (duración en segundos).

2. **Cruce de Datos sin Llave Foránea Física (`INNER JOIN`)**:
   * El operador **`INNER JOIN`** conecta las tablas sincronizándolas por su marca de tiempo común (`trips.start_ts = weather_records.ts`). Al ser una unión interna estricta, el motor omite automáticamente todos los registros de viajes para los cuales no existan reportes del clima asociados en esa hora.

3. **Cortes de Segmentación en el Dataset (`WHERE`)**:
   * **`pickup_location_id = 50` y `dropoff_location_id = 63`**: Restringe geográficamente los viajes para incluir únicamente las rutas que partieron desde Loop con destino final al Aeropuerto O'Hare.
   * **`EXTRACT(DOW FROM ...) = 6`**: Extrae el día de la semana a partir de la marca de tiempo de inicio. El valor numérico `6` representa rigurosamente el día **sábado** en el calendario SQL, aislando el comportamiento del fin de semana.

4. **Clasificación del Output (`ORDER BY`)**:
   * **`ORDER BY trips.trip_id`**: Ordena secuencialmente los resultados por la llave primaria operativa de los viajes para asegurar consistencia en el set de datos final.

### Prueba estadística de la hipótesis: Impacto climático en la duración de los viajes
Una vez saneada y consolidada la estructura del dataset `df_trips_weather`, procedemos a la fase analítica central del proyecto: la validación estadística de la hipótesis que plantea que **"la duración promedio de los viajes desde el Loop hasta el Aeropuerto Internacional O'Hare cambia los sábados lluviosos"**. El objetivo de esta prueba es determinar si los factores meteorológicos adversos (clasificados de forma binaria como *Good* o *Bad*) generan un impacto real y estadísticamente significativo en los tiempos de traslado operativos de Zuber, o si las variaciones observadas en la muestra se deben puramente al azar. Para ello, se separarán las observaciones en dos grupos de datos independientes y se aplicará un contraste paramétrico fundamentado, estableciendo un nivel de confianza estándar para guiar la toma de decisiones comerciales de la empresa.

````python

# 1. Separar las duraciones de los viajes en los dos grupos climáticos
duration_good = df_trips_weather[df_trips_weather['weather_conditions'] == 'Good']['duration_seconds']
duration_bad = df_trips_weather[df_trips_weather['weather_conditions'] == 'Bad']['duration_seconds']

print("=== 1. VERIFICACIÓN DE TAMAÑO DE GRUPOS ===")
print(f"Viajes con buen clima (Good): {len(duration_good)}")
print(f"Viajes con mal clima (Bad): {len(duration_bad)}\n")

# 2. Ejecutar la prueba de Levene para verificar la igualdad de varianzas
levene_stat, levene_p = stats.levene(duration_good, duration_bad)

print("=== 2. PRUEBA DE LEVENE (IGUALDAD DE VARIANZAS) ===")
print(f"Estadístico Levene: {levene_stat:.4f}")
print(f"p-value de Levene: {levene_p:.4f}")

# Determinar dinámicamente el parámetro equal_var según el p-value de Levene
if levene_p > 0.05:
    var_iguales = True
    print("-> Resultado: El p-value > 0.05 indica que las varianzas SON IGUALES (Se usará equal_var=True)\n")
else:
    var_iguales = False
    print("-> Resultado: El p-value <= 0.05 indica que las varianzas SON DIFERENTES (Se usará equal_var=False)\n")

# 3. Ejecutar la prueba T de Student para muestras independientes
t_stat, t_p = stats.ttest_ind(duration_good, duration_bad, equal_var=var_iguales)

print("=== 3. PRUEBA T DE STUDENT (PRUEBA DE HIPÓTESIS) ===")
print(f"Estadístico t: {t_stat:.4f}")
print(f"p-value del t-test: {t_p}")
````

#### Descripción y justificación del procedimiento seguido y presentación de resultados

#### Planteamiento de las hipótesis
Para evaluar el impacto de las condiciones climáticas adversas en el servicio de Zuber, definimos formalmente los enunciados estadísticos:

* **Hipótesis nula ($H_0$):** La duración promedio de los viajes desde el Loop hasta el Aeropuerto Internacional O'Hare los sábados lluviosos es igual a la duración promedio en los sábados de buen clima.
* **Hipótesis alternativa ($H_1$):** La duración promedio de los viajes desde el Loop hasta el Aeropuerto Internacional O'Hare los sábados lluviosos cambia (es diferente) en comparación con los sábados de buen clima.

#### Criterio estadístico utilizado y justificación
Se seleccionó la **Prueba t de Student para dos muestras independientes** (`scipy.stats.ttest_ind`) debido a los siguientes motivos metodológicos:

1. **Naturaleza de los grupos:** Se comparan dos muestras completamente separadas y autónomas entre sí (los viajes del grupo con clima "Good" frente a los del grupo con clima "Bad").
2. **Variable de estudio:** La variable dependiente analizada es el tiempo (`duration_seconds`), la cual es de tipo cuantitativa continua.
3. **Validación de Varianzas (Prueba de Levene - Iteración 2):** A diferencia de la entrega anterior, se incorporó formalmente la Prueba de Levene para evaluar la homogeneidad de varianzas. El resultado arrojó un **$p$-value de 0.8195** (superior al umbral estándar de 0.05), lo que demuestra científicamente que las varianzas de ambos grupos son estadísticamente **iguales**. Por lo tanto, se justificó rigurosamente la configuración del parámetro **`equal_var=True`** en la prueba $t$, abandonando la corrección de Welch.

#### Explicación de los resultados y conclusión comercial del proyecto
Al ejecutar el análisis estadístico con el dataset completamente depurado (865 registros libres de duplicados y valores en cero), se obtuvieron los siguientes resultados analíticos objetivos:
* **Estadístico $t$:** $-5.2915$
* **Valor p (p-value):** $1.54 \times 10^{-7}$ $(0.000000154)$

**Decisión estadística:** Dado que el valor $p$ ($1.54 \times 10^{-7}$) es críticamente menor que el nivel de significancia establecido ($\alpha = 0.05$), se cuenta con la evidencia matemática absoluta para **rechazar la hipótesis nula ($H_0$)** en favor de la hipótesis alternativa ($H_1$).

**Conclusión comercial:** La hipótesis de la investigación es correcta. Las condiciones meteorológicas desfavorables (lluvia o tormentas) los días sábados impactan y alteran de forma estadísticamente significativa la duración promedio de los viajes desde el centro (Loop) hasta el aeropuerto (O'Hare). El estadístico $t$ negativo refleja que los tiempos de viaje se elevan significativamente cuando el clima es adverso. 

A nivel estratégico, este incremento en la duración representa retrasos operativos causados por la congestión vial y la conducción preventiva. Zuber debe asimilar estos hallazgos dentro de su modelo de negocio para:
